```python
import zipfile
import os

# Define the path to the zip file
zip_file_path = '/content/archive.zip'

# Define the directory to extract to
extract_dir = '/content/extracted_data'
os.makedirs(extract_dir, exist_ok=True)

# Extract the zip file
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

print(f"Contents of '{zip_file_path}' extracted to '{extract_dir}':")
for item in os.listdir(extract_dir):
    print(item)
```

```python
import pandas as pd

# Find the first CSV file in the extracted directory and load it
csv_file = None
for root, _, files in os.walk(extract_dir):
    for file in files:
        if file.endswith('.csv'):
            csv_file = os.path.join(root, file)
            break
    if csv_file: # Break outer loop if a CSV is found
        break

if csv_file:
    print(f"Loading data from: {csv_file}")
    df = pd.read_csv(csv_file)
    print("First 5 rows of the DataFrame:")
    display(df.head())
else:
    print("No CSV files found in the extracted archive.")
```

```python
# Check for missing values
print("Missing values per column:")
display(df.isnull().sum())

print("\nTotal missing values in the DataFrame:")
display(df.isnull().sum().sum())
```

```python
# Check for duplicate rows
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {num_duplicates}")

if num_duplicates > 0:
    print("\nExamples of duplicate rows:")
    display(df[df.duplicated()].head())
```

### 1. Check for Missing Values

First, let's confirm if there are any missing values in each column of your DataFrame. This helps us understand the completeness of our data.

```python
# Calculate the number of missing values for each column
# `isnull()` returns a boolean DataFrame indicating missing values
# `sum()` then counts the `True` values (missing values) for each column
missing_values = df.isnull().sum()
display(missing_values)

# Calculate the total number of missing values in the entire DataFrame
# We sum again to get a single number for all missing entries
total_missing_values = missing_values.sum()
print(f"\nTotal missing values in the DataFrame: {total_missing_values}")
```

### 2. Check for Duplicates

Next, we'll check if there are any identical rows (duplicates) in the dataset. Duplicate entries can skew our analysis.

```python
# `duplicated()` returns a boolean Series, True for all duplicate rows (keeping the first occurrence as False)
# `sum()` counts how many True values there are, giving us the number of duplicate rows
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {num_duplicates}")

if num_duplicates > 0:
    print("\nHere are the first 5 examples of duplicate rows:")
    # Displaying the first few actual duplicate rows if any exist
    # `df[df.duplicated()]` filters the DataFrame to show only the duplicate rows
    display(df[df.duplicated()].head())
```

### 3. Show Basic Statistics for the 'Score' column

Understanding the distribution of the 'Score' column is crucial. We'll look at its mean, minimum, and maximum values.

```python
# Select the 'Score' column and apply the `describe()` method
# `describe()` provides a summary of central tendency, dispersion, and shape of the distribution
# It includes count, mean, standard deviation, min, max, and quartiles
print("Basic statistics for the 'Score' column:")
display(df['Score'].describe())
```

### 4. Find the Top 10 Happiest Countries

Let's identify the countries with the highest happiness scores. This involves sorting the DataFrame by 'Score' in descending order.

```python
# Sort the DataFrame by the 'Score' column in descending order (`ascending=False`)
# `head(10)` then selects the first 10 rows from the sorted DataFrame
print("Top 10 Happiest Countries:")
top_10_happiest = df.sort_values(by='Score', ascending=False).head(10)
display(top_10_happiest[['Country or region', 'Score', 'Overall rank']])
```

### 5. Find the Bottom 10 Happiest Countries

Conversely, we'll find the countries with the lowest happiness scores by sorting the DataFrame in ascending order.

```python
# Sort the DataFrame by the 'Score' column in ascending order (default for `ascending=True`)
# `head(10)` selects the first 10 rows from this sorted DataFrame
print("Bottom 10 Happiest Countries:")
bottom_10_happiest = df.sort_values(by='Score', ascending=True).head(10)
display(bottom_10_happiest[['Country or region', 'Score', 'Overall rank']])
```

### 6. Calculate Correlations with 'Score'

Understanding the relationship between 'Score' and other numerical columns (like 'GDP per capita', 'Social support', etc.) can reveal important insights. We'll calculate the Pearson correlation coefficient, which measures the linear relationship between two variables.

```python
# Calculate the correlation matrix for all numerical columns
# `numeric_only=True` ensures only numerical columns are considered
# Then, select the 'Score' column from the correlation matrix
# `sort_values(ascending=False)` arranges the correlations from highest to lowest
print("Correlation of each numerical feature with 'Score':")
correlations = df.corr(numeric_only=True)['Score'].sort_values(ascending=False)
display(correlations)
```
